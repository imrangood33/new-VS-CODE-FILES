### Dictionary

Dictionaries are used to store data values in key:value pairs.

A dictionary is a collection which is ordered*, changeable and does not allow duplicates.

Dictionaries are written with curly brackets, and have keys and values:

- **Ordered**: Items have a defined order, and that order will not change.
- **Unordered**: Items do not have a defined order, you cannot refer to an item by using an index.
- **Changeable**: Dictionaries are changeable, meaning that we can change, add or remove items after the dictionary has been created.
- **Duplicates Not Allowed**: Dictionaries cannot have two items with the same key.


````python
thisdict= {"brand": "Ford", "model": "Mustang", "year": 1964}
print(thisdict)
print(thisdict["brand"])
print(thisdict.get("model"))
````

## While loops and for loops

A while loop executes a block of statements as long as a condition is true. A for loop iterates over items of a sequence (like list, tuple, string, dict, or range).

### while loop
- Syntax:
```python
while condition:
    # body
```
- Example:
```python
i = 0
while i < 5:
    print(i)
    i += 1
```
- Notes: beware infinite loops; use `break` to exit early and `continue` to skip to the next iteration. A `while` loop can have an `else:` block that runs if the loop completes normally (no `break`).

### for loop
- Syntax:
```python
for item in iterable:
    # body
```
- Examples:
```python
# iterate over a list
for fruit in ["apple", "banana", "cherry"]:
    print(fruit)

# using range
for i in range(3):  # 0,1,2
    print(i)

# enumerate to get index and value
for idx, val in enumerate(["a", "b", "c"], start=1):
    print(idx, val)
```
- Iterating dictionaries:
```python
for key, value in {"a": 1, "b": 2}.items():
    print(key, value)
```
- Notes: `break`, `continue`, and `else` work similarly to `while` loops.

### Common tips
- Use `for` when you know the sequence or number of iterations; use `while` when you depend on a condition that may change in the loop body.
- Keep loop bodies simple and ensure the loop condition will eventually become false (for `while`) to avoid infinite loops.

````python
i=1
while i<6:
    print(i)
    i +=1
````

````python
# exit the loop when i is 3
i = 0
while i < 6:
    if i == 3:
        break
    print(i)
    i += 1
````

````python
# skip when i == 3
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
````

````python
# while else condition
i = 1
while i < 6:
    print(i)
    i += 1
else:
    print("i is no longer less than 6")
    
````

## For loop

A `for` loop iterates over the items of a sequence (list, tuple, string, dict, range, etc.).

Syntax:
```python
for item in iterable:
    # body
```

Examples:

- iterate a list
```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

- use the existing variable `i` (i == 6) to iterate a range
```python
# uses i defined earlier in the notebook
for n in range(i):
    print(n)
```

- get index and value with enumerate
```python
for idx, val in enumerate(fruits, start=1):
    print(idx, val)
```

- use break and continue
```python
for n in range(10):
    if n == 5:
        break     # exit loop
    if n % 2 == 0:
        continue  # skip even numbers
    print(n)
```

When to use: prefer `for` when you know the sequence or number of iterations; use `while` when looping depends on a changing condition.

````python
for x in "banana":
    print(x)
````

````python
#break statement can break loop early
fruits= ["apple", "banana", "cherry", "kiwi", "mango"]
for x in fruits:
    print(x)
    if x == "cherry":
        break
````
